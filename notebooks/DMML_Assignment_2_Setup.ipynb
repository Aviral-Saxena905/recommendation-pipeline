```python
import os

PROJECT_DIR = "/content/drive/MyDrive/recommendation-pipeline"

print("Project directory configured:", os.path.exists(PROJECT_DIR))
```

```python
schema_code = '''
import pandera.pandas as pa
from pandera.pandas import Column, Check

ratings_schema = pa.DataFrameSchema({
    "userId": Column(int, nullable=False),
    "movieId": Column(int, nullable=False),
    "rating": Column(
        float,
        Check.in_range(0.5, 5.0),
        nullable=False
    ),
    "timestamp": Column(int, nullable=False)
}, strict=True)

users_schema = pa.DataFrameSchema({
    "userId": Column(int, nullable=False, unique=True),
    "age": Column(int, Check.in_range(18, 64), nullable=False),
    "occupation": Column(str, nullable=False),
    "location": Column(str, nullable=False)
}, strict=True)
'''

schema_path = os.path.join(
    PROJECT_DIR, "src", "validation", "schema.py"
)

with open(schema_path, "w") as file:
    file.write(schema_code)

print("Initial validation schemas saved successfully!")
```

```python
additional_schemas_code = '''

movies_schema = pa.DataFrameSchema({
    "movieId": Column(int, nullable=False, unique=True),
    "title": Column(str, nullable=False),
    "genres": Column(str, nullable=False)
}, strict=True)

tags_schema = pa.DataFrameSchema({
    "userId": Column(int, nullable=False),
    "movieId": Column(int, nullable=False),
    "tag": Column(str, nullable=False),
    "timestamp": Column(int, nullable=False)
}, strict=True)

links_schema = pa.DataFrameSchema({
    "movieId": Column(int, nullable=False, unique=True),
    "imdbId": Column(int, nullable=False),
    "tmdbId": Column(float, nullable=True)
}, strict=True)
'''

with open(schema_path, "a") as file:
    file.write(additional_schemas_code)

print("All five dataset schemas are now defined!")
```

```python
with open(schema_path, "r") as file:
    print(file.read())
```

```python
validation_code = '''
import pandas as pd

from schema import (
    ratings_schema,
    users_schema,
    movies_schema,
    tags_schema,
    links_schema
)


def validate_dataset(dataframe, schema, dataset_name):
    try:
        validated_data = schema.validate(dataframe)

        print(f"{dataset_name}: VALIDATION PASSED")

        return validated_data

    except Exception as error:
        print(f"{dataset_name}: VALIDATION FAILED")
        print(error)

        return None
'''

validation_path = os.path.join(
    PROJECT_DIR,
    "src",
    "validation",
    "validate_data.py"
)

with open(validation_path, "w") as file:
    file.write(validation_code)

print("Validation script created successfully!")
```
